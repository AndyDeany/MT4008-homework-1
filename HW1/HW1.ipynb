### Data Science Example

As a simple example, I will use the *planets* dataset from the seaborn library.

```python
import seaborn as sns
import matplotlib.pyplot as plt


planets = sns.load_dataset("planets")
planets.head()  # Display the first few rows
```

Above we see the format of the data. Let's plot planets' masses against their orbital periods.

```python
sns.scatterplot(data=planets, x="mass", y="orbital_period")
plt.title("Planetary mass vs orbital period")
plt.xlabel("mass")
plt.ylabel("orbital period")
plt.show()
```

The data seems to span several orders of magnitude,
so a few planets with very high values for mass and/or orbital period
make it diffucult to see the data from the planets with lower values properly.
Let's use a log/log scale instead.

```python
sns.scatterplot(data=planets, x="mass", y="orbital_period")
plt.xscale("log")
plt.yscale("log")
plt.title("Planetary mass vs orbital period (log/log scale)")
plt.xlabel("log(mass)")
plt.ylabel("log(orbital period)")
plt.show()
```

This is much better, and we can see the distribution of planets more clearly.
